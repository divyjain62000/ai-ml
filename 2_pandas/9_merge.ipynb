```python
import numpy as np
import pandas as pd
```

```python
# Need to find orange cap holder of season: means need to find batsman who hit most run in season
#Problem: we have matches dataset but it does not have batsman run and we have delivery dataset but it does not have season
match=pd.read_csv('matches.csv')
match.head(2)
```

```python
delivery=pd.read_csv('deliveries.csv')
delivery.head(2)
```

```python
# In both the data set we have one common thing id or match id we will merge both dataset based on that
new=delivery.merge(match,how='inner',left_on='match_id',right_on='id')
```

```python
# We have done groupby on 2 things first on season and inside each season on batsman
season_wise_batsman=new.groupby(['season','batsman'])['batsman_runs'].sum().sort_values(ascending=False)
season_wise_batsman
```

```python
type(season_wise_batsman)
```

```python
# Now we want to conver season_wise_batsman to data frame above series has 2 index one is season and other is batsman
#to convert it into data frame we will use rest_index function
season_wise_batsman_df=season_wise_batsman.reset_index()
season_wise_batsman_df
```

```python
# As we want to find orange cap holder from each season means only one batsman from each season who hit most of the run 
# so we will drop all remaining batsman of that season and keep only first one
top_batsman_data=season_wise_batsman_df.drop_duplicates(subset=['season'],keep='first').sort_values('season')
top_batsman_data
```
