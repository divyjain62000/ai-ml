# Dropping missing data using dropna()

```python
data.shape
```

```python
new_data=data.dropna()
```

```python
new_data.shape
```

```python
# drop row in which all column value is None or missing, so here we don't have any such column
data.dropna(how='all').shape
```

```python
# bydefgault axis=0 so it drop row If we changed it to 1 then it start dropping column
data.dropna(axis=1).shape
```

```python
# bydefgault axis=0 so it drop row If we changed it to 1 then it start dropping column and we can use 'how' here also
data.dropna(axis=1,how='all').shape
```

```python
data.dropna(subset=['Cabin']).shape
```

```python
data.dropna(subset=['Embarked']).shape
```

```python
# Its kind of OR operation
data.dropna(subset=['Cabin','Embarked']).shape
```
