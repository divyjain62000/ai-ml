```python
import numpy as np
import pandas as pd
```

```python
delivery=pd.read_csv('deliveries.csv')
```

```python
delivery.head(2)
```

```python
# To find team against which viratl kohli hit most of the runs
mask=delivery['batsman']=='V Kohli'
vk=delivery[mask]
```

```python
vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(5)
```

```python
def run_scored(batsman_name):
    filtered_data=delivery[delivery['batsman']==batsman_name]
    return filtered_data.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(1).index[0]
```

```python
run_scored('MS Dhoni')
```
