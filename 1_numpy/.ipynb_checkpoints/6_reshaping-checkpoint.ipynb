# Stacking: combine two array

```python
arr1
```

```python
arr2=np.arange(6,12).reshape(2,3)
arr2
```

```python
# Combine array horizontally
np.hstack((arr1,arr2))
```

```python
# Combinen array vertically
np.vstack((arr1,arr2))
```

# Spliting

```python
arr3=np.arange(1,13).reshape(4,3)
arr3
```

```python
#horizontal split
np.hsplit(arr3,3)
```

```python
#vertical split
np.vsplit(arr3,2)
```

```python
np.vsplit(arr3,4)
```
