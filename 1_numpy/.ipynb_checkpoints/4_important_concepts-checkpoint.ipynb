```python
import numpy as np
```

```python
arr1=np.arange(24)
arr1
```

```python
arr2=np.arange(24).reshape(6,4)
arr2
```

# Slicing

```python
#Print 2nd column
arr2[:,1]
```

```python
#Print 2nd and 3rd column
arr2[:,1:3]
```

```python
#Print 9,10,13,14
arr2[2:4,1:3]
```

```python
#Print 18,19,22,23
arr2[4:,2:]
```

# Iteration

```python
for i in arr2:
    print(i)
```

```python
#To print items
for i in np.nditer(arr2):
    print(i)
```

# Fancy Indexing

```python
arr2
```

```python
# To get 0,2,4 index rows
arr2[[0,2,4]]
```

```python
# To get 1,3 index columns
arr2[:,[1,3]]
```
