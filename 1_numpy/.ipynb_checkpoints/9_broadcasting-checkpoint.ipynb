# Bradcasting

```python
import numpy as np
```

```python
a1=np.arange(8).reshape(2,4)
a2=np.arange(8,16).reshape(2,4)
print(a1)
print("-"*10)
print(a2)
```

```python
a1+a2
```

```python
a3=np.arange(9).reshape(3,3)
a4=np.arange(3).reshape(1,3)
print(a3)
print("-"*10)
print(a4)
```

```python
a3+a4
```

```python
a5=np.arange(3).reshape(3,1)
a6=np.arange(9).reshape(3,3)
print(a5)
print("-"*10)
print(a6)
```

```python
a5+a6
```

```python
a7=np.arange(3).reshape(1,3)
a8=np.arange(3,6).reshape(3,1)
print(a7)
print("-"*10)
print(a8)
```

```python
a7+a8
```

```python
a9=np.arange(1,2).reshape(1,1)
a10=np.arange(8).reshape(4,2)
print(a9)
print("-"*10)
print(a10)
```

```python
a9+a10
```
