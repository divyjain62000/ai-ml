# Filtering or Indexing with boolean arrays

```python
import numpy as np
```

```python
arr=np.random.randint(low=1,high=100,size=20).reshape(4,5)
arr
```

```python
arr>50
```

```python
# To get final array in which number are greater than 50
filtered_arr=arr[arr>50]
filtered_arr
```

```python
# To get final array in which number are greater than 50 and odd
filtered_arr=arr[(arr>50) & (arr%2!=0)]
filtered_arr
```

```python
# In array wherever number are greater than 0 and odd make them 0
arr[(arr>50) & (arr%2!=0)]=0
arr
```
