# Dot product

```python
arr1=np.arange(6).reshape(2,3)
arr2=np.arange(6,12).reshape(3,2)
```

```python
print("arr1\n",arr1)
print("arr2\n",arr2)
print("dot product\n",arr1.dot(arr2))
```
